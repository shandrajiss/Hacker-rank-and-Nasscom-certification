```python
# Invalid: variable name contains a space
# car name = 20

# Invalid: variable name contains a dollar sign
# car$ = 5

# Invalid: variable name contains a hyphen
# my-car = 10

# Invalid: variable name starts with a number
# 2car = 20

# Invalid: variable name contains a space and number
# car 2s = 20

# Valid: number appears after the first character
car2 = 20
print(car2)

# Valid: number appears at the end
bike3 = 30
print(bike3)

# Valid: variable name starts with an underscore
_car = 40
print(_car)

# Valid: variable contains multiple underscores
my_new_car = 50
print(my_new_car)

# Invalid: variable name contains an exclamation mark
# car! = 10

# Invalid: variable name contains a percent sign
# car% = 10

# Invalid: variable name contains a dot
# car.name = 10
```

```python
# Valid: variable name uses uppercase letters
CAR = 200
print(CAR)
```

```python
# Valid: variable name uses mixed case
myCar = 300
print(myCar)
```

```python
# Variable names are case-sensitive
car = 10
Car = 20
print(car)
print(Car)
```
